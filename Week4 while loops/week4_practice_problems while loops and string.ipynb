Remember in week 2, we asked you to write a function that takes an integer input and returns an integer equal to the number of digits it has? In this week's we ask you to write the function that does the same but use a loop that you just learned in class. The function should work with all integers representable on your computer.   
For example, say your input number is 123456, the output of the function should then be 6.



````python
def count_digits(num_signed):
    '''(int) -> int
    
    Returns the number of digits of an interger.
    
    ## Example
    
    >>> count_digits(37621)
    5 

    >>> count_digits(-468)
    3

    >>> count_digits(0)
    0
    '''
    # Initialize: digit = 0, convert negative numbers to its absolute value
    num = abs(num_signed)
    digit = 0

    # Set the number to compare with
    max_str = '1'
    max = int(max_str)

    # conpare the number with the maximum number in one specific digit range
    while num >= max:
        max_str = max_str + '0'
        max = int(max_str)
        digit = digit + 1

    return digit

````

Also in week 2, we asked you to write a function that takes an integer as input and returns that integer flipped. Unlike Week 2, the code must work with any integer and a single digit flipped is the number itself. If a number ends in a zero, the flipped number will not have a leading 0. Some examples: 
* 2 -> 2
* 34 -> 43
* 123450 -> 54321

As above, you should complete this question using a you just learned in class. 


````python
def flipper(num):
    # Fill in the docstring and your code here
    '''(int) -> int
    
    Filps the number, without the tailing 0.
    
    ## Example
     
    >>> flipper(47386384)
    
    48368374'''
    # initialize -> define the divider
    divider = 1

    # find the digits of the number
    digit = count_digits(num)
    count = 0

    # initialize the flipped number
    flipped_number_str = ''

    # divide the number by ten times the devider, and find the first digit of remainder
    while count < digit:
        ten_times_divider = divider * 10
        remainder = num % ten_times_divider

        # we need only the first digit of the remainder
        first_digit_of_remainder = remainder // divider

        first_digit_of_remainder_str = str(first_digit_of_remainder)



        flipped_number_str = flipped_number_str + first_digit_of_remainder_str

        # prepare to go to next loop
        divider = divider * 10

        count = count + 1

    flipped_number = int(flipped_number_str)
    return flipped_number

    
````

````python
def sum_of_odd(num): 
    # Fill in the docstring and your code here
    '''(int) -> int
    
    Takes in an odd number and returns the sum of odd numbers from 1 to the number provided.
    
    ## Example
     
    >>> sum_of_odd(5)
    9
    
    >>> sum_of_odd(19)
    100'''
    sum = num
    # num must be greater than one; otherwise we will have a +(-1) at the end 
    # and get one smaller than we want
    while num > 1:
        num = num - 2
        sum = sum + num

    return sum
````
