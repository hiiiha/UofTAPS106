
**Hint:** For truncating, there are mutiple ways to achieve this. Think about using `int`, or alternatively, you can also think about using the [`math`](https://docs.python.org/3/library/math.html) module.

````python
def truncate_round(num):
    '''(float) -> float
    
    Truncate and round a float number to one decimal place.
    '''
    # Remember to write the docstring here. Make it a habit so that you don't lose point in the upcoming exams.
    # Truncate a number
    num10 = 10 * num
    num10int = int(num10)
    num_truncated = num10int / 10

    # round a number
    num_rounded = round(num, 1)

    print('Truncated to one decimal place:', num_truncated, '\nRounded to one decimal place:', num_rounded)
    
````

# Question 4

Write a  trigonometry function. Ask a user for an angle, specified in **degrees**. Then print the sine, cosine, and tangent of that angle. Round to 3 decimal places. For example:
```
Enter an angle in degrees:  60
sin(60.0) is 0.866
cos(60.0) is 0.5
tan(60.0) is 1.732
```

Printing should occur **inside** the function. You can use [`math`](https://docs.python.org/3/library/math.html) module, where applicable.

````python
def trig_calc(angle):
    '''(number) -> number

    Asks for an angle in degrees and returns the sine, cosine, and tangent of the angle input, rounded to 3 decimal places.'''
    
    import math
    # Convert degrees to radians
    angle_radian = math.radians(angle)

    # Calculate the sin, cos, and tan
    sin_noround = math.sin(angle_radian)
    cos_noround = math.cos(angle_radian)
    tan_noround = math.tan(angle_radian)

    # Round the numbers
    sin = round(sin_noround,3)
    cos = round(cos_noround,3)
    tan = round(tan_noround,3)

    # Print the results
    print('sin(' + str(angle) + ')', 'is', sin, '\ncos(' + str(angle) + ')', 'is', cos, '\ntan(' + str(angle) + ')', 'is', tan,)

````

Test the fucntion in the cell below. **Fill in any missing code, where applicable.**

````python
user_input = input("Enter an angle in degrees: ")
print('Angle selected is', user_input + '°')
angle = float(user_input)
trig_calc(angle)
````
