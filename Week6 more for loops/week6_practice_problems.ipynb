## Question 1
Question 1 of week 5 asked you to swap cases of a string input and print the result. For example:
```
Please input a string with upper and lower case letters:  aBc
AbC
```
Perhaps you used a `while` loop to solve that questions. For this question, slightly modify you code to use `for` loop instead.


````python
# Write you code here
def swap_case(original: str):
    '''(str) -> str

    Swaps the character cases in the original string.'''
    newstr = ''
    for char in original:
        if char.lower() == char:  # char is in lower case, convert to upper
            new_char = char.upper()
            newstr += new_char
        elif char.upper() == char: # char is in upper, convert to lower
            new_char = char.lower()
            newstr += new_char
    
    return newstr

original = input('input string here')
print(swap_case(original))
````

## Question 2
Question 2 of week 5 asked you to remove exclamation and question marks from an input string. For example,
```
Example:

Input: Hello!, Output: Hello
Input: How are you?, Output: How are you
```
For this question, slightly modify you code to use `for` loop instead.

````python
# Write you code here
def remove_exclamation_question_mark(original: str):
    '''(str) -> str
    
    Removes the exclamation and question marks from the original string.'''
    new_str = ''
    for char in original:
        if char != '?' and char != '!':   # make sure no ? and ! go in the new_str
            new_char = char
            new_str = new_str + new_char

    return new_str

to_be_removed = input('input something')
print(remove_exclamation_question_mark(to_be_removed))
````

## Question 3

You are given two pieces of data:
1. A string `s` of length $n$.
2. An integer $k$, where $k$ is a factor of $n$.

Because $n$ is divisible by $k$, we can split string `s` into $\frac{n}{k}$ sub-strings where each segment consists of a contiguous block of $k$ characters. 

Write a program that gets the user to enter a string and an integer $k$ and prints the segments in order but have the characters in each segment reversed and separate the segments by space. Example: 
```
Please enter the string: University
Please enter k: 2
nU vi re is yt
```

```
Please enter the string: Hello!
Please enter k: 3
leH !ol 
```
It is a good idea to check if $k$ does indeed divide $n$ evenly. If it doesn't print an error message and exit. If it does, do the splitting and reversing as described above.

*Hint:* Break the problem into three steps: 1. Get the input and check. 2. Break the string into $\frac{n}{k}$ substrings. 3. Reverse each sub-string and print it. You might find the `range()` function helpful. You should write at least one function to do this. 

*Another hint:* The main algorithm for doing this (e.g. after the input, etc) is only a few lines of string manipulation.



````python
# Write you code here
def check_divisible(dividend: int, divider: int):
    '''(int, int) -> bool
    
    Checks if the dividend is divisible by the divider, returns true if divisible.'''
    if dividend % divider == 0:
        return True
    else:
        return False
    
def split_str_reverse_print(to_be_split: str, substring_length: int):
    '''(str, int) -> str
    
    Split the string into substrings of substring_length, then reverse each substring. 
    
    Returns the reversed substrings added up.
    
    ## Example
    >>> split_str_reverse_print(University, 2)
    nU vi re is yt
    '''
    pieces = len(to_be_split) // substring_length
    i = 0
    final_str = ''
    while i < pieces:  # make sure it does not exceed the index available
        split_str = to_be_split[i * substring_length:(i+1) * substring_length] # the substring
        reverse_str = split_str[::-1] + ' '
        final_str += reverse_str
        
        i = i + 1

    return final_str

# input
inputstr = input('the string you want to manipulate:')
input_k = int(input('the k you want'))

# checks if the k is divisible
if check_divisible(len(inputstr), input_k):
    print(split_str_reverse_print(inputstr, input_k))
else:
    print('This is not a valid k (cannot divide n evenly)')
````

## Question 4
Write a function that takes in a string of any length as a parameter and prints out a table that specifies the number of times each vowel (a,e,i,o,u) appears. Your function should take be able to identify a vowel whether it is upper or lower-case.

1. Write the function using the `count()` method.
2. Write the function without using the `count()` method. (Hint: nested loop)


````python
def count_aeiou_using_count_method(string: str):
    string = string.lower() # transform to all lower case

    for i in range(5):
        occur = string.count('aeiou'[i])
        print('aeiou'[i] + ':', occur, 'times')

count_aeiou_using_count_method('aeiouAaaerr')
````

````python
def count_aeiou_using_nested_loop(string: str):
    string = string.lower() # transform to all lower case
    
    for i in range(5):
        # this block of code is the same as the count method
        occur = 0
        for char in string: 
            if char == 'aeiou'[i]:
                occur += 1

        print('aeiou'[i] + ':', occur, 'times')

count_aeiou_using_nested_loop('aeiouAaaerrty')
````

## Question 5
Question 5 of week 5 asked you to write a function that takes in a DNA strand and produces its matching counterpart in a double helix. Again, here are two paired strands as an example:

    ATGGGCAATCGATGGCCTAATCTCTCTAAG
    TACCCGTTAGCTACCGGATTAGAGAGATTC
Modify what you did in week 5 but this time use a loop rather than string methods.

````python
def matching_dna(strand: str):
    '''(str) -> str
    
    Takes a DNA strand string and returns the matching counterpart in a double helix.'''
    strand = strand.upper()
    matching = ''
    for char in strand:
        if char == 'A':
            matching += 'T'
        if char == 'T':
            matching += 'A'
        if char == 'G':
            matching += 'C'
        if char == 'C':
            matching += 'G'

    return matching


print(matching_dna('ATGGGCAATCGATGGCCTAATCTCTCTAAG'))
````
