## Question 1
Ask the user for a size (`size`) and a maximum value (`max_value`). Write a program that randomly assigns values between 0 and `max_value` to a `size`-by-`size` matrix. **The matrix should be represented as a nested tuple (i.e. a tuple of tuples).** Print the matrix to the screen, and then calculate the sum of values around the "outside" (border) of the matrix: that is, the sum of the entries in the first and last rows and first and last columns but without double counting the corner entries. For example, if the random matrix is:

    8 7 2 0 4
    8 4 0 2 5
    1 9 4 6 5
    0 2 6 9 3
    1 7 3 9 5

then the sum around the outside is 68. You will need to import the random module and use an appropriate function for generating the numbers.




````python
def generate_matrix(size: int, max: int):
    '''(int, int) -> tuple
    
    Takes integers `size` and `max` to generate a matrix represented in a tuple of `size` * `size` and the `max` as the maximum value in the matrix.'''
    # use random module to generate integers
    import random

    # define an empty list for holding the matrix
    matrix_list = []
    
    # loop through "size" time to generate every row
    for every_row in range(size):

        # define an empty row list for holding a row of the matrix
        matrix_row_list = []

        # loop through to generate every value, for "size" times
        for every_value in range(size):
            value = random.randint(0, max)   # random generate a 
            matrix_row_list.append(value)    # append the value generated

        matrix_row_tuple = tuple(matrix_row_list)  # now one row is generated, convert to tuple
        matrix_list.append(matrix_row_tuple)       # append the row to the matrix list
    matrix_tuple = tuple(matrix_list)          # after all loops, convert matrix to tuple

    return matrix_tuple

def outer_sum(matrix: tuple):
    '''(tuple) -> int
    
    Takes a matrix represented by a tuple and returns the sum of the values around the "outside" of the matrix. Assumes square matrix.'''
    # verify the tuple
    #for row in matrix:
    #    if type(row) != tuple:
    #        return -1
    #    for value in row:
    #        if type(value) != int:
    #            return -1
    size_rows = len(matrix)
    #size_columns = len(matrix[0])
    #if size_rows != size_columns:
    #    return -1
    
    # first, get the size of the matrix
    size = size_rows
    
    # size is 1, just return itself
    if size == 1:
        return int(matrix[0][0])
    
    # get the first row and last row added
    first_row_sum = sum(matrix[0])
    last_row_sum = sum(matrix[size - 1])

    # get the column sum from the second to the second last row
    first_col_sum = 0
    last_col_sum = 0

    for i in range(1, size - 1):
        first_col_sum += matrix[i][0]         # add the first value
        last_col_sum += matrix[i][size - 1]   # add the last value

    # add all up
    allsum = first_row_sum + last_row_sum + first_col_sum + last_col_sum
    return allsum
````

````python
mat = generate_matrix(99,33)
print(mat)
print(outer_sum(mat))
````

## Question 2
Write a program that asks the user for a matrix size (call it, `n`) and then randomly assigns the values 1 to `n`<sup>2</sup> to a `n` x `n` matrix, **not repeating any values!** Again, the matrix should be represetned as a nested tuple.

Then print the matrix to the screen.  For example:

      3 25 12  4 19
      5 17 21 10  6
      1  8 16  2  7
     13 22 24 11 20
     23 14 15  9 18

Notice that you'll need some way to make sure there are no repeated numbers. 

If you do this in a simple, but inefficient way, you’ll notice it takes longer and longer to place the next value or find the next location. That’s OK for a 5 x 5 matrix but won’t be appropriate for a much larger (say, 1000 x 1000) matrix. So then what? Try writing a version that’s efficient, that takes the same amount of time to place the first number as the last. Maybe you can think of more than two different ways?

Bonus: Investigate the Python `time` module and time your different implementations.


````python
def non_repeating_matrix(n: int): # n is the size
    '''(int) -> tuple
    
    Generate a n*n matrix in a tuple with no repeating values. All values are within [1, n^2].'''
    import random

    matrix_list_one_line = list(range(1, n ** 2 + 1))
    
    random.shuffle(matrix_list_one_line)
    
    # split the one line list    
    matrix_list = []

    for row_index in range(0, n):
        matrix_row = matrix_list_one_line[5 * row_index : 5 * row_index + 5]
        matrix_row_tuple = tuple(matrix_row)
        matrix_list.append(matrix_row_tuple)

    matrix_tuple = tuple(matrix_list)

    return matrix_tuple
````

````python
import time
start = time.process_time_ns()

print(non_repeating_matrix(5000))

end = time.process_time_ns()
print('time (nsec)', end - start)
````
