## Question 1

Create a class named `DecimalConvert`. The object should have an attribute  called `decimal`.  Then, write a class method (call it `convert`) to convert the positive base-10 integer (the `decimal` attribute) to binary. The class method should return a string of 0s and 1s corresponding to the binary value of the value of the integer. You may have to do a bit of research to figure out how to convert a base-10 to integer to binary. The class method should just return `None` if the input is not a positive integer.

Then outside the class definition, write a program to prompt the user for input, then create an instance of `DecimalConvert` to convert the user input to binary, and print the results. 

````python
class DecimalConvert:
    '''Stores a decimal number in the .decimal attribute.'''
    def __init__(self, decimal: int) -> None:
        self.decimal = decimal

    def convert(self) -> str | None:
        '''() -> str
        
        Returns the decimal number stored as a binary representation'''
        if self.decimal < 0:
            return None
        if self.decimal == 0:
            return '0'
        converting = self.decimal
        binary_str = ''
        while converting != 0:
            remainder = converting % 2
            converting = converting // 2
            binary_str = str(remainder) + binary_str

        return binary_str
````
