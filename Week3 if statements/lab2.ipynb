```python
##############################
# APS106 Winter 2024 - lab2  #
# Circle overlap             #
##############################


def circle_overlap(circ1_centre_x, circ1_centre_y, circ1_radius,
                    circ2_centre_x, circ2_centre_y, circ2_radius):
    """
    (int, int, int, int, int, int) -> str

    Function determines whether two circles overlap. When circles
    overlap, the function checks for the following scenarios
        1. The two circles perfectly overlap
        2. The first circle is contained within the second
        3. The second circle is contained within the first
        4. The circle have overlapping area, but neither is completely
           contained within the other
    
    Function inputs represent x and y coordinates circle centres and their
    radii (see lab document)
           
    The function returns a string describing the overlap scenario.
    
    # Examples

    >>> circle_overlap(0,1,3,6,4,1)
    'no overlap'
    
    >>> circle_overlap(0,1,3,0,1,3)
    'identical circles'
    
    >>> circle_overlap(1,1,10,6,7,1)
    'circle 2 is contained within circle 1'
    
    >>> circle_overlap(-1,-2,2,0,0,11)
    'circle 1 is contained within circle 2'
    
    >>> circle_overlap(1,-2,2,-4,0,5)
    'circles overlap'
    """

    import math

    # General structure:
    # no overlap: the distance between two circle's centre is greater than their radius added.
    # contain: the distance between two circle's centre is smaller than the radius difference.
    # other cases: overlap

    distance_centres = math.sqrt((circ1_centre_x - circ2_centre_x) ** 2 + (circ1_centre_y - circ2_centre_y) ** 2)
    radius_sum = circ1_radius + circ2_radius
    radius_diff = abs(circ1_radius - circ2_radius)
    
    # identical circles
    if (circ1_centre_x == circ2_centre_x) and (circ1_centre_y == circ2_centre_y) and (circ1_radius == circ2_radius):
        return ('identical circles')

    else: # identical circles already sorted out
        # no overlap: the distance between two circle's centre is greater than their radius added.
        if distance_centres >= radius_sum: # 
            return ('no overlap')

        # contain: the distance between two circle's centre is smaller than the radius difference.
        elif distance_centres <= radius_diff:
            # compare radius. Should not have same radius because could be identical circles
            if circ1_radius < circ2_radius: 
                return ('circle 1 is contained within circle 2')
            if circ1_radius > circ2_radius: # similar, does not need elif because they are already mutually exclusive
                return ('circle 2 is contained within circle 1')

        # overlap: all other cases than no overlap or contain
        else:
            return ('circles overlap')
                


```

```python
print(circle_overlap(0,1,3,0,1,3)) # identical
print(circle_overlap(0,1,3,6,4,1)) # no overlap
circle_overlap(1,1,10,6,7,1) # 2 in 1
circle_overlap(-1,-2,2,0,0,11) # 1 in 2
circle_overlap(1,-2,2,-4,0,5) # overlap only

# my testcases
circle_overlap(0, 0, 3, 2, 0, 1) # 2 in 1
circle_overlap(0, 0, 3, 4, 0, 1) # no overlap
circle_overlap(0, 0, 1, 0, 0, 2) # 1 in 2
circle_overlap(0, 0, 3, 2, 3, 1) # overlap?
print(circle_overlap(0, 0, 1, 0, 1, 1)) # overlap
```
