## Question 1

Write a function that takes the role of a store clerk. 

Ask a user to enter two floating point numbers: the *cost* of an item and the *amount of money* remitted to pay for the item.  
Then respond appropriately: calculate the change due to the customer *or* ask the customer for more money. 

For example:
```
Cost of the item: 3.56
Amount tendered: 5.00
Change: 1.44
```
or
```
Cost of the item: 3.56
Amount tendered: 3.00
Still due: 0.56
```
**Note:** print out the change and amount due to two decimal places

````python
# Complete the function in this cell
def clerk_calc(cost, tendered):
    '''(float, float) -> float
    
    Takes the total cost of the item and the amount paid, then calculate the change or still due amount.

    Prints out the cost of the item, amount tendered, and change or still due amount. 
    
    Returns the change as a positive number and still due amount as a negative number.
    '''
    # Print the cost and amounttendered
    print('Cost of the item:', cost)
    print('amount tendered:', tendered)

    # change to cent and integer to avoid imprecision
    cost_cent = int(100 * cost)
    tendered_cent = int(100 * tendered)

    # Calculate the change or amount due and print
    change_cent = tendered_cent - cost_cent
    change = change_cent / 100

    # Print the results
    if change > 0:
        print('Change:', change)
    elif change < 0:
        print('Still due:', -change)
    else:
        print('None due or change')
    return change
    
````

````python
# Test your function here. 
cost = 0.2
tender = 0.3
clerk_calc(cost, tender)
````

## Question 3
You may have already realized that Canada  no longer have pennies in circulation. Therefore, let's write a function that round a number to the nearest nickels.

For example, <span>\$</span>1.53   is rounded to <span>\$</span> 1.55, <span>\$</span>2.47 is rounded to <span>\$</span>2.45, <span>\$</span>4.42 is rounded to <span>\$</span>4.40.

The function should take in the change as the input as return the rounded as the output

````python
def round_to_nickel(change):
    '''
    (float) -> float
    
    Round the dollar amount to the nearest nickels.

    ## Examples

    >>> round_to_nickel(1.53)
    1.55

    >>> round_to_nickel(1.78)
    1.8

    >>> round_to_nickel(2.07)
    2.05
    '''
    change_cents = int(100 * change)

    change_times_2 = change_cents * 2
    change_times_2_rounded = round(change_times_2, -1)
    change_rounded = change_times_2_rounded / 2

    change_rounded_dollars = change_rounded / 100
    return change_rounded_dollars
````

````python
# You should test the function before proceeding to the next step
print(round_to_nickel(1.53))
print(round_to_nickel(1.78))
print(round_to_nickel(2.07))
````

Redo Question 5 from Week 2 but using an if-statement. Write a function that takes a positive integer input less than 100000 and returns an integer corresponding to the number of digits in the number. Use this function in a program that prompts the user for an integer and prints the number of digits in the user’s input. 

````python
def count_digits(num):
    '''(int) -> int
    
    Calculate and return the number of digits of a number less than 100,000.
    
    Parameters
    -------
    num : int
          The number you want to calculate the number of digits.

    Returns
    -------
    int
        The number of digits of (num).

    Examples
    -------
    >>> count_digits(3727)
    4
    '''
    print('the number you entered is', num)
    if num == 100000:
        return 6
    elif num < 100000 and num >= 10000:
        return 5
    elif num < 10000 and num >= 1000:
        return 4
    elif num < 1000 and num >= 100:
        return 3
    elif num < 100 and num >= 10:
        return 2
    elif num < 10 and num >= 1:
        return 1
    else:
        print("please enter a positive number less than 100000.")
````

````python
# Test your function here
user_input = input("Please input a number less than 100,000: ")
num  = int(user_input)
print(count_digits(num))
````
